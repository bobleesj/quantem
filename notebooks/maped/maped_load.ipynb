# Load a saved MAPED merge

Opens a result saved by [maped_automated.ipynb](maped_automated.ipynb). Edit `SAVE_PATH`. Works on CUDA and on a Mac (MPS); the file loads onto whichever GPU the machine has.


```python
# Install once. The widget brings quantem.gpu with it; MAPED lives in quantem.
# pip install --extra-index-url https://test.pypi.org/simple "quantem-gpu>=0.0.1rc11" git+https://github.com/bobleesj/quantem.widget.git
# pip install git+https://github.com/bobleesj/quantem.git@main
```

```python
from quantem.gpu import io
from quantem.widget import Show4DSTEM

SAVE_PATH = "/path/to/results/merged_master.h5"

merged = io.load(SAVE_PATH)
```

```python
Show4DSTEM(merged)
```

## Selected diffraction patterns

Three scan positions, read from the saved file; indexing returns a Torch tensor.


```python
from quantem.core.visualization import show_2d

positions = [(int(merged.shape[0] * f), int(merged.shape[1] * f)) for f in (0.25, 0.5, 0.75)]
fig, ax = show_2d([merged[row, col] for row, col in positions], title=[f"scan ({row}, {col})" for row, col in positions], cmap="inferno", norm="power_sqrt")
```
