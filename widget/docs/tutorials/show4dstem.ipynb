# Show4DSTEM

`Show4DSTEM` opens a 4D-STEM dataset with live virtual detectors: a bright-field / annular-dark-field aperture over the diffraction stack on one side, the resulting virtual image on the other. It accepts a NumPy array, a PyTorch tensor, a quantem `Dataset4dstem`, or the output of `load(...)`.

```{note}
This example is exported with `offline=True`: for a small dataset the full uint16 stack ships to the browser and the virtual-detector math runs in **WebGPU**, so the widget below is fully interactive with no Python kernel. Drag the bright-field disk or change the detector radius and the virtual image recomputes in the browser (bit-exact to the kernel, sub-millisecond). Large datasets keep the kernel path.
```

````python
import numpy as np

# Synthetic 4D-STEM: a bright-field disk whose center shifts with scan position
# (a DPC-like field), so the virtual detectors produce a meaningful image.
scan, det = 32, 64  # small + light: 8 MB offline stack, browser-friendly
qr, qc = np.mgrid[0:det, 0:det]
radius = 16.0
data = np.zeros((scan, scan, det, det), dtype=np.uint16)
for sr in range(scan):
    for sc in range(scan):
        cr = det / 2 + 10.0 * (sr / scan - 0.5) * 2
        cc = det / 2 + 10.0 * (sc / scan - 0.5) * 2
        disk = ((qr - cr) ** 2 + (qc - cc) ** 2) <= radius ** 2
        data[sr, sc] = (disk * 1000).astype(np.uint16)
data.shape, data.dtype
````

## Virtual detectors

The bright-field disk shifts across the scan, so the default BF virtual image shows the DPC-like contrast.

````python
from quantem.widget import Show4DSTEM

Show4DSTEM(data, sampling=(0.2, 0.2, 1.0, 1.0), units=['A', 'A', 'mrad', 'mrad'],
           title='Synthetic 4D-STEM', offline=True)
````
