# DPC / iCoM on MacBook (MPS) - no binning

Center-of-mass, DPC, and iDPC computed on **Apple Metal**, on the **full no-bin**
4D-STEM stack (512x512x192x192). The CoM runs as a raw-Metal kernel over chunked
uint16 buffers - no float cast, fits 24 GB, ~2s. Rotation + iDPC are small-field.

Bit-exact vs quantem.live's CUDA DPC.


## 1. Load (no bin) + compute DPC


```python
from quantem.widget import load, dpc, Show2D

data = load("/Users/macbook/data/samsung/logic_013_master.h5")          # no-bin MPS chunks
result = dpc(data)                  # Metal CoM -> rotation -> iDPC (~6s no-bin)
print(f"rotation {result.rotation_deg:.1f} deg, phase {result.phase.shape}, {result.elapsed:.1f}s")
```

## 2. iDPC phase  (integrated - the projected potential / atoms)


```python
Show2D(result.phase)
```

## 3. Raw CoM / DPC field  (the gradient - x and y deflection)


```python
Show2D(result.com_col)   # CoM_x
```

```python
Show2D(result.com_row)   # CoM_y
```
